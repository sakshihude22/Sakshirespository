$Case-3$: **range(start,stop,step)**
- if any one provide 3 values
- first value is a **start**
- second value is a **stop**
- third value is a **step**
- **here step plays very important role**
    - step means gap
    - in above two cases the step size by default one only
- for example step a **positive value**
    - start = start
    - last = stop-1
    - increament by + step
- for example step a **negative value**
    - start=start
    - last = stop+1
    - decrement by step size

```python
for i in range(2,20,-2):
    print(i, end=' ')
```

**Counter wrapper**

- if we want count success ones
- if we want count desired outputs
- then go for counter wrapper
- similar to sum wrapper, but its updated by+1
    - count=0
    - count=count+1

```python
count=0
for i in range(1,76):
    if 75%i==0:
        count+=i
print(count)
```
