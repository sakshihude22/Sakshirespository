$case-2$: **for loop with if condition**

```python
l=[10,15,19,20,22,27]
out=[i for i in l if i%2==0]
out
```

```python
input=['hyd','bengaluru','mum#bai','chen#nai']
out=[]
for i in input:
    if '#' in i:
        out.append(i)
out
```

```python
input=['hyd','bengaluru','mum#bai','chen#nai']
out=[i for i in input if '#' in i]
out
```

```python
input=['hyd','bengaluru','mum#bai','chen#nai']
out=[i for i in input if '#' not in i]
out
```
